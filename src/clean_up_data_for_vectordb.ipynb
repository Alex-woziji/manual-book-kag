```python
import pandas as pd 
```

```python
df = pd.read_excel("../builder/classified.xlsx") 
```

```python
df['category'].value_counts()
```

```python
#failure type (1)  --> 1-3类 --> 3类里面任选6个solution
#Spindle & Drive\Mechanical Components\Automation & Robotics

#failure type (2)  --> 4-6类 --> 3类里面任选5个solution
#Coolant & Lubrication\Electrical & Control\Hydraulic & Pneumatic
```

```python
failure_type_1 = ["Spindle & Drive",'Mechanical Components','Automation & Robotics']
failure_type_2 = ["Coolant & Lubrication",'Electrical & Control','Hydraulic & Pneumatic']
```

```python
df.loc[df['category'].isin(failure_type_1),'failure_type'] = "1"
```

```python
df.loc[df['category'].isin(failure_type_2),'failure_type'] = "2"
```

```python
for ind,row in df.iterrows():
    with open(f"../builder/classfired_failure_type_txt/type_{str(row['failure_type'])}/{row['file_name']}","w",encoding="utf-8") as f:
        f.write(row['content'])
```

```python
df 
```

```python
failure_type_1_solution = []
for i in df.loc[df['failure_type'] ==  "1","solutions"]:
    failure_type_1_solution.extend(eval(i)) 
```

```python
failure_type_2_solution = []
for i in df.loc[df['failure_type'] ==  "2","solutions"]:
    failure_type_2_solution.extend(eval(i)) 
```

```python
len(failure_type_1_solution)
```

```python
len(failure_type_2_solution)
```

```python
import random
random.seed(42)

failure_type_1_solution_sample = random.sample(failure_type_1_solution, 6)
failure_type_2_solution_sample = random.sample(failure_type_2_solution, 6)
```

```python
import json

solution_mapping = {"failure_type_1_solution_sample":failure_type_1_solution_sample,
 "failure_type_2_solution_sample":failure_type_2_solution_sample}

with open('../builder/solution_mapping.json', 'w', encoding='utf-8') as f:
    json.dump(solution_mapping, f, ensure_ascii=False, indent=4)
```
